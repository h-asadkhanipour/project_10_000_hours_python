```python
import os
import json
import openpyxl
from datetime import datetime, timedelta
```

```python
# load json file
with open('../reports_data/daily_reports.json') as file:
    json_file = json.load(file)
```

```python
mydic = dict(json_file)
```

```python
mydic
```

```python
# load excel file
excel_file = openpyxl.load_workbook('../reports_data/daily_reports.xlsx')
excel_sheet = excel_file['TenThousandHoursReport']
row_counter = 2
for date in mydic:
    if mydic[date] != None:
        study_hours = mydic[date]['study_work_hours']
        explanation = mydic[date]['explanation']
        excel_sheet[f'B{row_counter}'] = study_hours
        excel_sheet[f'C{row_counter}'] = explanation
        row_counter += 1

excel_file.save('../reports_data/daily_reports.xlsx')
```
